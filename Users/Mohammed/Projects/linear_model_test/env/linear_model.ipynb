## Linear Model Training

```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
data = pd.read_csv('data/ecommerce_data.csv')
data
```

## EDA on our dataset

```python
data.info()
```

```python
sns.heatmap(data.isna()) ## checking for any null values in our dataset
```

```python
data.describe()
```

```python
data ## picked only continuous variables for our linear model training
```

```python
sns.pairplot(data,kind='scatter', plot_kws = {'alpha':0.5}) ## pairplot to check correlation between every predictor with the target varaibles
```

```python
X = data[['Avg. Session Length','Time on App','Time on Website','Length of Membership']] 
y = data['Yearly Amount Spent']
```

# Training the Model

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 42) ## split our dataset into training and testing samples
X_train.shape
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
y_predict = model.predict(X_test)
```

## Model Evaluation

```python
sns.scatterplot(x=y_predict, y=y_test) ##Evaluating our model
plt.xlabel('Predictions')
plt.title('Evaluation of our Linear Model')
```

```python
sns.scatterplot(y_predict - y_test)
```

```python
sns.histplot(y_predict-y_test, kde = True, bins=30) ## our residuals are normally distributed 
```

```python
r2 = r2_score(y_test, y_predict) 
print(f'R2 score: {r2*100:.2f}%')
```

```python
## Our model predicts the actual data with an r2_score of 0.98
## As the score nears 1, the better the model predicts the actual value
```
